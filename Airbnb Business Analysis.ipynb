## Data Analysis 

```python
df['price'].describe()
```

```python
import seaborn as sns
price_analysis =sns.histplot(data=df,x='price')
```

```python
price_analysis
```

```python
plt.figure(figsize=(10,5))

sns.histplot(data=df, x='price', bins=50)

plt.title("Price Distribution of New York Airbnb Listings")
plt.xlabel("Price")
plt.ylabel("Number of Listings")

plt.show()
```

```python
# checking any outliers in price column
sns.boxplot(data=df,x='price')
```

```python
df = df[df['price'] < 1500]
```

```python
sns.boxplot(data=df,x='price')
```

```python
plt.figure(figsize=(10,5))

sns.histplot(data=df, x='price', bins=50)

plt.title("Price Distribution of New York Airbnb Listings")
plt.xlabel("Price")
plt.ylabel("Number of Listings")

plt.show()
```

```python
df.columns
```

```python
room_type_analysis = df.groupby('room_type', as_index=False)['price'].sum()

plt.figure(figsize=(10, 5))

sns.barplot(data=room_type_analysis, x='room_type', y='price')

plt.title("Total Price by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Total Price")

plt.show()
```
